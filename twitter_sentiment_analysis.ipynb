```python
import pandas as pd
from textblob import TextBlob
import matplotlib.pyplot as plt
```

```python
twitter_data = pd.read_csv("training.1600000.processed.noemoticon.csv", encoding='latin-1', header=None)
```

```python
twitter_data.columns = ['sentiment', 'id', 'date', 'query', 'user', 'tweet']
```

```python
twitter_data.head()
```

```python
def analyze_sentiment(text):

    analysis = TextBlob(str(text))

    if analysis.sentiment.polarity > 0:
        return "Positive 😊"

    elif analysis.sentiment.polarity < 0:
        return "Negative 😡"

    else:
        return "Neutral 😐"
```

```python
sample_data = twitter_data.head(500)
```

```python
sample_data['Predicted_Sentiment'] = sample_data['tweet'].apply(analyze_sentiment)
```

```python
sample_data[['tweet', 'Predicted_Sentiment']].head()
```

```python
sample_data['Predicted_Sentiment'].value_counts()
```

```python
sample_data['Predicted_Sentiment'].value_counts().plot(kind='bar')

plt.title("Twitter Sentiment Analysis")
plt.xlabel("Sentiment")
plt.ylabel("Number of Tweets")

plt.show()
```
