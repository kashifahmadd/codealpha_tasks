```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
netflix_data = pd.read_csv("netflix_titles.csv")
```

```python
sns.countplot(x='type', data=netflix_data)
plt.title("Movies vs TV Shows on Netflix")
plt.show()
```

```python
netflix_data['rating'].value_counts().plot(kind='bar')
plt.title("Context Rating Distribution")
plt.xlabel("Ratings")
plt.ylabel("Count")
plt.show()
```

```python
netflix_data['country'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Countries Producing Netflix Content")
plt.xlabel("Country")
plt.ylabel("Number of shows")
plt.show()
```

```python
plt.hist(netflix_data['release_year'])
plt.title("Distribution of Release Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

# Conclusion

1. Netflix contains more movies than TV shows.

2. Certain countries contribute significantly more Netflix content.

3. TV-MA and TV-14 are among the most common ratings.

4. Most Netflix content was released in recent years.
