```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
youtube_data = pd.read_csv("Global YouTube Statistics.csv", encoding='latin-1')
```

```python
youtube_data.head()
```

```python
youtube_data.info()
```

```python
youtube_data.isnull().sum()
```

```python
top_channels = youtube_data.sort_values(by='subscribers', ascending=False).head(10)

plt.figure(figsize=(10,5))

sns.barplot(x='subscribers', y='Youtuber', data=top_channels)

plt.title("Top 10 YouTubers by Subscribers")

plt.show()
```

```python
youtube_data['channel_type'].value_counts().head(10).plot(kind='bar')

plt.title("Top Channel Categories")

plt.xlabel("Channel Type")
plt.ylabel("Count")

plt.show()
```

```python
youtube_data['Country'].value_counts().head(10).plot(kind='pie', autopct='%1.1f%%')

plt.title("Top Countries with YouTube Channels")

plt.show()
```

```python
plt.hist(youtube_data['video views'])

plt.title("Distribution of Video Views")

plt.xlabel("Views")
plt.ylabel("Frequency")

plt.show()
```

```python
sns.scatterplot(x='uploads', y='subscribers', data=youtube_data)

plt.title("Uploads vs Subscribers")

plt.show()
```
